# Practical 1: Study of deep learning packages (viva only)

**The idea.** **Keras** is the easy front end (`Sequential`, `Dense`, `compile`, `fit`). **TensorFlow** does the maths underneath on CPU/GPU. **PyTorch** is the other big library and is popular in research. **Theano** is the retired ancestor. Every `keras.` line in practicals 2–6 is Keras running on TensorFlow.

| Package | What it is | Remember |
|---|---|---|
| **TensorFlow** | Google's library for tensor maths + training (CPU/GPU/TPU) | Production and deployment (TF Lite, Serving). TF2 runs eagerly; `tf.function` compiles a graph |
| **Keras** | High-level API to define, compile, fit, evaluate models | Ships inside TF as `tf.keras`. Easiest to learn |
| **PyTorch** | Meta's library with a dynamic graph and autograd | Graph built as Python runs, so you can debug line by line |
| **Theano** | Early library (Univ. of Montreal) that compiled maths graphs | Development stopped in 2017 |

## A tensor and autograd, seen in 3 lines
A **tensor** is a multi-dimensional array. **Autograd** records the operations and works out gradients automatically.

```python
x = tf.Variable(3.0)
with tf.GradientTape() as tape:
    y = x ** 2            # y = x^2
print("dy/dx at x=3:", tape.gradient(y, x).numpy())   # 2x = 6
```

## Viva answers
**Q1. Keras vs TensorFlow?** Keras is a high-level API for defining, compiling, fitting and evaluating models. TensorFlow is the library that does the tensor maths underneath, on CPU or GPU. In current versions Keras ships inside TensorFlow as `tf.keras`.

**Q2. PyTorch's dynamic graph vs TensorFlow's graph?** PyTorch builds the computation graph as the Python code runs, so you can debug line by line. TensorFlow traditionally built the graph first and then ran it (faster, easier to deploy). TF2 runs eagerly by default but can still compile graphs with `tf.function`.

**Q3. What does autograd do?** It records the operations done on tensors and computes the gradients automatically, by backpropagation through that graph.

**Q4. What is a tensor?** A multi-dimensional array of numbers: a number is 0-D, a list 1-D, a table 2-D, and an image batch like (60000, 28, 28, 1) is 4-D.
